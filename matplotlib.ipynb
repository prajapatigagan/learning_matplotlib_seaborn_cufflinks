## Matplotlib Learing

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

####   Matplotlib basics

```python
x=np.linspace(0,5,11)
print("X",x)
y=x**2
print("Y",y)
```

```python
plt.plot(x,y)
plt.title("Hello")
plt.xlabel("X values")
plt.ylabel("Y values")
```

#### Subplot in Matplotlib

```python
plt.subplot(2, 2, 1)
plt.plot(x,y)
plt.plot(y,x)
plt.subplot(2, 2, 2)
plt.plot([1, 2, 3], [1, 4, 9])
plt.subplot(2, 2, 3)
plt.bar([1, 2, 3], [5, 3, 7])
plt.subplot(2, 2, 4)
plt.scatter([1, 2, 3], [5, 2, 8])
```
